### 1.2. Por que a Ciência de Dados para Desenvolvedores Web?

Tradicionalmente, o desenvolvimento web é guiado por requisitos, especificações e, por vezes, intuição. A ciência de dados introduz uma mudança de paradigma fundamental: da construção de funcionalidades baseadas em suposições para a criação de experiências orientadas por dados e evidências. Para um desenvolvedor web, essa habilidade abre um leque de possibilidades que vai muito além da simples implementação de *features*.

A relevância se manifesta em cenários práticos e cotidianos:

* **Análise de Logs de Servidor:** Em vez de usar comandos como `grep` para caçar erros em arquivos de log gigantescos, um desenvolvedor pode carregar esses logs em um DataFrame do Pandas. Isso permite agregar dados, identificar padrões de erro recorrentes, encontrar gargalos de performance que afetam múltiplos usuários e até detectar atividades suspeitas, como um aumento súbito de requisições de um mesmo IP, que poderiam indicar uma tentativa de ataque.
* **Otimização da Experiência do Usuário (UX):** A intuição sobre o que constitui uma boa UX pode ser validada ou refutada com dados. Analisando dados de cliques, mapas de calor, tempo de permanência em páginas e fluxos de navegação, é possível identificar com precisão onde os usuários estão encontrando dificuldades, quais botões são ignorados e quais caminhos levam a uma maior taxa de conversão. Essa análise informa diretamente as decisões de design e desenvolvimento, tornando-as mais eficazes.
* **Fundamentação de Testes A/B:** Um desenvolvedor web frequentemente implementa as duas versões de uma página para um teste A/B. Com habilidades em ciência de dados, seu papel se expande. Ele pode não apenas implementar, mas também analisar os resultados do teste, aplicando rigor estatístico para determinar qual versão é a vencedora e por qual margem. Isso o coloca no centro das decisões que impactam diretamente as métricas de negócio, como engajamento e receita.
* **Personalização de Conteúdo e Sistemas de Recomendação:** Gigantes da tecnologia como Netflix e Amazon construíram seus impérios sobre a personalização. A base para isso é a análise de dados. Ao analisar o histórico de compras, visualizações ou interações de um usuário, é possível construir *features* que recomendam produtos, artigos ou conteúdos relevantes, aumentando drasticamente o engajamento, a retenção e o valor gerado por cada usuário.
* **Desenvolvimento de *Features* Inteligentes:** A ciência de dados é o motor por trás de funcionalidades que parecem "mágicas". Um formulário de cadastro que se adapta e simplifica com base no perfil do usuário, um sistema de busca que aprende com as pesquisas anteriores para oferecer resultados mais relevantes, ou um dashboard administrativo que, em vez de apenas mostrar números, destaca os insights mais importantes para o gestor. Todas essas são aplicações diretas da análise de dados no ciclo de desenvolvimento.

A incorporação da ciência de dados no desenvolvimento web representa uma evolução do papel do desenvolvedor. Em vez de ser um executor de tarefas definidas por outros, ele se torna um parceiro estratégico. A mentalidade muda de reativa ("construir o que foi pedido") para proativa ("propor o que os dados sugerem"). Quando um gerente de produto sugere mudar a cor de um botão, o desenvolvedor com visão de dados pode ir além e propor um experimento controlado para medir o impacto real da mudança na conversão. Ele passa a dialogar com as áreas de produto, marketing e negócios usando a linguagem universal e objetiva dos dados, influenciando diretamente os resultados da empresa.

## Seção 2: O Mapa da Mina: Navegando em um Projeto de Dados com CRISP-DM

### 2.1. Desmistificando a Ciência de Dados

Antes de mergulhar na prática, é essencial estabelecer um modelo mental claro sobre o que é a ciência de dados. Formalmente, é um campo interdisciplinar que utiliza métodos científicos, processos, algoritmos e sistemas para extrair conhecimento e insights de dados estruturados e não estruturados. Na prática, ela se apoia em três pilares fundamentais:

1.  **Conhecimento de Domínio/Negócio:** A compreensão do contexto do problema que se está tentando resolver.
2.  **Matemática e Estatística:** O ferramental teórico para quantificar relações, testar hipóteses e construir modelos.
3.  **Ciência da Computação/Programação:** As habilidades para coletar, armazenar, processar e analisar grandes volumes de dados.

A boa notícia para desenvolvedores web é que um desses pilares — a programação — já é uma área de domínio. O desafio, e o objetivo deste curso, é construir as pontes para os outros dois pilares, focando nas ferramentas e conceitos mais relevantes para agregar valor em projetos web.

### 2.2. O Ciclo de Vida de um Projeto de Ciência de Dados: CRISP-DM

Assim como o desenvolvimento de software tem suas metodologias (como Agile, Scrum, Waterfall), os projetos de ciência de dados também se beneficiam de um processo estruturado. A metodologia mais amplamente adotada e comprovada pela indústria é o **CRISP-DM**, sigla para *Cross-Industry Standard Process for Data Mining*. Ele fornece um roteiro que organiza o projeto em seis fases distintas, garantindo que nenhuma etapa crítica seja negligenciada.

![Diagrama do CRISP-DM](https://upload.wikimedia.org/wikipedia/commons/thumb/b/b9/CRISP-DM_Process_Diagram.png/400px-CRISP-DM_Process_Diagram.png)

As seis fases do CRISP-DM são:

1.  **Entendimento do Negócio (Business Understanding):** Esta é a fase inicial e, indiscutivelmente, a mais crítica. Antes de tocar em qualquer linha de código, é preciso entender profundamente o objetivo do projeto do ponto de vista do negócio. Perguntas como "Qual problema estamos tentando resolver?", "Como o sucesso será medido?" e "Qual o impacto esperado para a empresa?" são respondidas aqui. Um erro nesta fase pode levar à construção de uma solução tecnicamente perfeita para o problema errado.
2.  **Entendimento dos Dados (Data Understanding):** Com os objetivos de negócio definidos, a atenção se volta para os dados. Esta fase envolve a coleta inicial de dados e uma primeira exploração para se familiarizar com eles. As atividades incluem identificar as fontes de dados, descrever seus atributos e realizar uma verificação inicial da qualidade.
3.  **Preparação dos Dados (Data Preparation):** Esta é a fase que consome mais tempo e será o foco principal da nossa aula. Ela abrange todas as atividades para construir o conjunto de dados final (o *dataset* limpo) a partir dos dados brutos. Isso inclui limpeza, integração de diferentes fontes, formatação e criação de novas variáveis.
4.  **Modelagem (Modeling):** Aqui é onde os algoritmos de machine learning e outras técnicas estatísticas são aplicados. Vários modelos podem ser treinados e ajustados para encontrar aquele que melhor responde à pergunta de negócio.
5.  **Avaliação (Evaluation):** Uma vez que um ou mais modelos foram construídos, eles precisam ser rigorosamente avaliados. A avaliação não é apenas técnica (ex: acurácia do modelo), mas também de negócio: o modelo realmente atende aos critérios de sucesso definidos na primeira fase?
6.  **Implantação (Deployment):** A fase final, onde o modelo é colocado em produção para gerar valor. Para um desenvolvedor web, isso pode significar integrar o modelo a uma aplicação existente, por exemplo, através de uma API que retorna uma predição em tempo real.

É crucial entender que o CRISP-DM não é um processo linear em cascata (waterfall). As setas no diagrama indicam um fluxo cíclico e iterativo. É comum e esperado retornar de uma fase posterior para uma anterior. Por exemplo, durante a avaliação, pode-se descobrir que o modelo não está performando bem, o que exige um retorno à fase de preparação de dados para criar novas *features*, ou até mesmo um retorno ao entendimento do negócio para redefinir o problema.

Essa natureza iterativa é, na verdade, um poderoso mecanismo de mitigação de riscos, muito alinhado com as filosofias ágeis que os desenvolvedores web conhecem bem. Em vez de seguir um plano rígido por meses para só no final descobrir uma falha fundamental, o CRISP-DM incentiva a validação e o refinamento contínuos. Ele garante que o projeto permaneça alinhado aos objetivos de negócio e se adapte às descobertas feitas ao longo do caminho, evitando o maior risco de todos: gastar tempo e recursos construindo a solução errada.

### 2.3. O Foco de Hoje: A Fase de Preparação dos Dados

A fase de preparação de dados é frequentemente a menos glamorosa, mas a mais impactante. Estima-se que cientistas de dados gastam de 50% a 80% do tempo de um projeto nesta etapa. A razão para isso é simples e encapsulada no ditado "Garbage In, Garbage Out" (Lixo Entra, Lixo Sai). A qualidade de qualquer análise, visualização ou modelo de machine learning é diretamente limitada pela qualidade dos dados que o alimentam. Dedicar tempo e rigor a esta fase é o que separa projetos de dados bem-sucedidos dos fracassados.

### 3.2. A Anatomia das Estruturas de Dados do Pandas

O poder do Pandas reside em suas duas estruturas de dados principais: a `Series` e o `DataFrame`.

* **Series:** É uma estrutura de dados unidimensional, análoga a uma única coluna em uma planilha ou um array do NumPy, mas com um diferencial importante: um **índice** associado que rotula cada elemento. Isso permite um acesso mais flexível e semântico aos dados.

```python
# Exemplo de uma Series com índice customizado
import pandas as pd
vendas_por_pais = pd.Series([500000, 450000, 300000], index=['UK', 'Germany', 'France'])
print(vendas_por_pais)
```

* **DataFrame:** É a estrutura de dados bidimensional e a principal ferramenta de trabalho em Pandas. Um DataFrame é, essencialmente, uma tabela. Ele pode ser visto como uma coleção de objetos `Series` que compartilham o mesmo índice. Cada coluna em um DataFrame é uma `Series`.

### 3.3. A Filosofia dos "Tidy Data" (Dados Organizados)

Ao trabalhar com dados, a forma como eles são organizados é tão importante quanto as ferramentas usadas para analisá-los. O conceito de **Tidy Data**, formalizado por Hadley Wickham, propõe um formato padrão para estruturar dados que facilita enormemente a análise e a modelagem. A filosofia é simples e se baseia em três regras inter-relacionadas:

1.  **Cada variável forma uma coluna.**
2.  **Cada observação forma uma linha.**
3.  **Cada tipo de unidade observacional forma uma tabela.**

Embora essas regras pareçam óbvias, a maioria dos dados encontrados no mundo real é considerada "messy" (bagunçada). Isso ocorre porque, frequentemente, os dados são organizados para facilitar a *entrada* de dados ou a *apresentação* em relatórios, e não a *análise* computacional. Grande parte do trabalho na fase de preparação de dados consiste em transformar dados "messy" em dados "tidy".

Para um desenvolvedor web, existe uma analogia poderosa: o formato "Tidy Data" funciona como uma **API padronizada para ferramentas de análise**. Desenvolvedores entendem o valor de APIs bem definidas e consistentes. Uma API mal projetada ou inconsistente leva a código frágil, retrabalho e frustração. Da mesma forma, dados "messy" são como uma API ruim para as ferramentas de análise. Uma função de plotagem pode esperar os dados em um formato, enquanto uma função de agregação espera em outro, forçando o analista a reformatar os dados constantemente.

O formato "Tidy Data" estabelece um contrato, uma interface consistente. Quando os dados aderem a este formato, as bibliotecas como Pandas, Seaborn, Matplotlib e Scikit-learn "sabem o que esperar". O código para criar um gráfico se torna intuitivo, como `seaborn.lineplot(data=df, x='ano', y='vendas')`, porque a API da biblioteca foi projetada para consumir dados nesse formato padronizado. Portanto, aprender sobre "Tidy Data" não é apenas uma questão de seguir uma regra acadêmica, mas de aplicar um princípio de design de software que promove a interoperabilidade e a eficiência no ecossistema de análise de dados.

## Seção 4: A Realidade dos Dados: Lidando com a Imperfeição (Teoria)

### 4.1. As Dimensões da Qualidade de Dados

"Dados sujos" é um termo genérico. Para abordar o problema de forma sistemática, a qualidade dos dados é avaliada através de várias dimensões. Compreender essas dimensões nos fornece um framework para auditar um conjunto de dados e identificar problemas específicos.

| Dimensão | Descrição | Exemplo no E-commerce | Fontes |
| :--- | :--- | :--- | :--- |
| **Completude** | Todos os dados necessários estão presentes? Faltam valores em campos importantes? | Uma transação não possui o `CustomerID`, impedindo a análise do comportamento daquele cliente. | [34, 35] |
| **Validade** | Os dados estão em um formato ou intervalo esperado? | Um `UnitPrice` é negativo, o que é fisicamente ou comercialmente impossível. | [34, 35] |
| **Precisão** | Os dados refletem corretamente o evento do mundo real que representam? | O preço de um produto no dataset é R\$ 10, mas no site ele custa R\$ 100. | [34, 35] |
| **Consistência** | Os dados não se contradizem dentro do mesmo dataset ou entre diferentes fontes? | O mesmo `StockCode` (ID do produto) aparece com duas descrições diferentes: 'caneca azul' e 'blue mug'. | [35] |

### 4.2. Identificando os "Vilões" da Qualidade de Dados

Com base nessas dimensões, podemos catalogar os problemas mais comuns que encontraremos na prática, os verdadeiros "vilões" da qualidade dos dados:

* **Dados Faltantes (Missing Data):** Representados em Pandas como `NaN` (Not a Number), `None` ou `NaT` (Not a Time). Eles podem surgir por diversas razões: falhas em sistemas de coleta, campos opcionais em formulários que não foram preenchidos, ou junções de tabelas onde não havia correspondência.
* **Dados Duplicados:** Linhas inteiras que são cópias exatas umas das outras. Se não forem tratadas, podem inflar métricas como contagens de vendas ou número de usuários, levando a conclusões erradas.
* **Tipos de Dados Incorretos:** Um problema extremamente comum. Números que são armazenados como texto (ex: `'1,234.56'`), datas armazenadas como strings (ex: `'10-12-2023'`), ou IDs numéricos que deveriam ser tratados como categorias (ex: `CustomerID`). Isso impede operações matemáticas e de data/hora.
* **Inconsistência de Formato:** Variações na forma como a mesma informação é registrada. Exemplos incluem diferenças de capitalização ('Brasil', 'brasil', 'BRASIL'), abreviações ('UK', 'United Kingdom'), ou múltiplos formatos de data ('DD/MM/AAAA' vs 'MM-DD-YYYY').
* **Outliers:** Valores que são drasticamente diferentes da maioria dos outros dados. Um outlier pode ser um erro de digitação (ex: idade de 150 anos) ou pode representar um evento real, mas raro (ex: uma única compra de valor extremamente alto). Eles podem distorcer severamente análises estatísticas como a média e precisam ser investigados com cuidado.

## Seção 5: Mãos à Obra: Limpeza de Dados com o Dataset de E-commerce

Agora, aplicaremos a teoria na prática, utilizando o Pandas para limpar nosso dataset de e-commerce. Este processo será o núcleo da nossa aula prática.

### 5.1. Carregando e Inspecionando os Dados (A Primeira Investigação)

O primeiro passo é carregar os dados em um DataFrame e realizar uma inspeção inicial para entender sua estrutura e identificar os primeiros sinais de problemas.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Carregar o dataset. A codificação 'ISO-8859-1' é necessária para ler caracteres especiais presentes no arquivo.
try:
    df = pd.read_csv('data.csv', encoding='ISO-8859-1')
except FileNotFoundError:
    print("Erro: Arquivo 'ecommerce_data.csv' não encontrado. Certifique-se de que ele está no diretório correto.")
    df = None # Define df como None para evitar erros nas células seguintes

if df is not None:
    # 1. Visualizar as primeiras 5 linhas
    print("Primeiras 5 linhas:")
    display(df.sample(5))

    # 2. Verificar as dimensões do DataFrame (linhas, colunas)
    print(f"\nDimensões do DataFrame: {df.shape}")

    # 3. Obter um resumo técnico do DataFrame
    print("\nInformações do DataFrame:")
    df.info()

    # 4. Obter estatísticas descritivas das colunas numéricas
    print("\nEstatísticas Descritivas:")
    display(df.describe())
```

A saída desses comandos já revela vários problemas:
* `df.info()` mostra que as colunas `Description` e `CustomerID` têm valores não nulos em menor quantidade que o total de entradas, indicando a presença de dados faltantes. Além disso, `InvoiceDate` é do tipo `object` (string), e não um tipo de data, o que impedirá análises temporais.
* `df.describe()` revela anomalias nas colunas numéricas: `Quantity` tem um valor mínimo negativo, o que parece ilógico para uma quantidade de produto. `UnitPrice` tem um valor mínimo de 0, o que também requer investigação.

### 5.2. Tratando Dados Faltantes (`CustomerID`)

Vamos quantificar os dados faltantes em cada coluna.

```python
if df is not None:
    # Contar valores nulos por coluna
    print("Contagem de valores nulos por coluna:")
    print(df.isnull().sum())
```

A saída confirma que `CustomerID` tem um número significativo de valores ausentes (mais de 135.000, o que representa quase 25% do total de registros). A decisão de como tratar esses dados faltantes não é puramente técnica, mas uma decisão com implicações de negócio. Uma abordagem ingênua seria simplesmente remover todas as linhas com `CustomerID` nulo.

No entanto, é preciso perguntar: *por que* esses dados estão faltando? No contexto de um e-commerce, a hipótese mais provável é que essas transações correspondam a compras feitas por "convidados" (*guest checkout*), ou seja, clientes que não fizeram login ou não possuem uma conta. Se removermos essas linhas, estamos descartando 25% de nossas transações e, mais criticamente, eliminando completamente nossa capacidade de analisar o comportamento de um segmento importante de clientes: os compradores não registrados.

Uma estratégia mais sofisticada poderia ser preencher esses valores nulos com um identificador especial, como `'GUEST'`, permitindo análises comparativas entre clientes registrados e convidados. Para os fins desta aula, focada na limpeza para uma análise de vendas por clientes *identificados*, procederemos com a remoção, mas é fundamental ter consciência do viés que essa decisão introduz na análise.

```python
if df is not None:
    # Removendo linhas onde CustomerID é nulo
    df.dropna(subset=['CustomerID'], inplace=True)

    # Verificando as dimensões e a contagem de nulos novamente
    print(f"\nNovas dimensões após remover nulos em CustomerID: {df.shape}")
    print(df.isnull().sum())
```

### 5.3. Eliminando Duplicatas

Registros duplicados podem inflar métricas e distorcer resultados. Vamos identificá-los e removê-los.

```python
if df is not None:
    # Verificar a quantidade de linhas duplicadas
    print(f"\nNúmero de linhas duplicadas: {df.duplicated().sum()}")

    # Remover as linhas duplicadas, mantendo a primeira ocorrência
    df.drop_duplicates(inplace=True)

    # Verificar as dimensões novamente
    print(f"Novas dimensões após remover duplicatas: {df.shape}")
```

O método `drop_duplicates()` remove linhas que são cópias exatas umas das outras. O parâmetro `inplace=True` modifica o DataFrame diretamente, sem a necessidade de reatribuí-lo a uma nova variável.

### 5.4. Corrigindo Tipos de Dados

Como vimos no `df.info()`, `InvoiceDate` e `CustomerID` precisam ter seus tipos corrigidos.

```python
if df is not None:
    # Converter 'InvoiceDate' para o tipo datetime
    df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

    # Converter 'CustomerID' para string. É uma boa prática tratar IDs como strings
    # para evitar que sejam usados em cálculos matemáticos indevidos.
    # O .0 é removido ao converter para int primeiro.
    df['CustomerID'] = df['CustomerID'].astype(int).astype(str)

    # Verificar os tipos de dados novamente
    print("\nTipos de dados após correção:")
    df.info()
```

A conversão de `InvoiceDate` para `datetime` é particularmente poderosa, pois nos permite extrair facilmente componentes como ano, mês, dia da semana e hora, que serão cruciais para análises futuras.

### 5.5. Padronizando Inconsistências Textuais (`Description`)

Dados textuais frequentemente sofrem com inconsistências de capitalização e espaços em branco. Vamos padronizar a coluna `Description`.

```python
if df is not None:
    # Remover espaços em branco no início e no fim da string e converter para minúsculas
    df['Description'] = df['Description'].str.strip().str.lower()
```

Os métodos do acessador `.str` no Pandas são extremamente úteis para aplicar operações de string a uma coluna inteira de forma vetorizada e eficiente.

### 5.6. Detectando e Lidando com Outliers e Casos Especiais

Agora, vamos investigar os valores anômalos que encontramos em `Quantity` e `UnitPrice`.

#### O Caso da `Quantity` Negativa
Vamos filtrar o DataFrame para ver as linhas com quantidade negativa.

Observando a saída, notamos que os `InvoiceNo` (números de fatura) associados a essas transações começam com a letra 'C', que, conforme a descrição do dataset, indica um **cancelamento**. Isso é um insight crucial: estes não são erros de dados, mas sim uma **regra de negócio representada nos dados**. São devoluções. Para uma análise focada em *vendas brutas*, devemos remover essas linhas. No entanto, para uma análise de *taxa de devolução* ou *churn de clientes*, essas linhas seriam o foco principal. Para nosso estudo de caso atual, vamos prosseguir removendo-as.

```python
if df is not None:
    # Manter apenas as linhas com quantidade positiva
    df = df[df['Quantity'] > 0]
    print(f"\nDimensões após remover cancelamentos: {df.shape}")
```

#### O Caso do `UnitPrice` Zero

#### Detectando Outliers Numéricos com o Método IQR
Mesmo após a limpeza, podem existir outliers legítimos que distorcem as análises. O método do Intervalo Interquartil (IQR) é uma técnica estatística robusta para identificá-los.

1.  **Visualização com Boxplot:** Um boxplot é uma excelente ferramenta para visualizar a distribuição e os outliers de uma variável.

```python
if df is not None:
    plt.figure(figsize=(10, 6))
    sns.boxplot(y=df['Quantity'])
    plt.title('Boxplot da Coluna Quantity')
    plt.show()
```

O gráfico mostrará claramente pontos que estão muito distantes do corpo principal dos dados.

2.  **Cálculo e Aplicação do IQR:**

```python
if df is not None:
    # Calcular Q1, Q3 e IQR para 'Quantity'
    Q1_qty = df['Quantity'].quantile(0.25)
    Q3_qty = df['Quantity'].quantile(0.75)
    IQR_qty = Q3_qty - Q1_qty

    # Definir os limites para detecção de outliers
    limite_inferior_qty = Q1_qty - 1.5 * IQR_qty
    limite_superior_qty = Q3_qty + 1.5 * IQR_qty

    # Fazer o mesmo para 'UnitPrice'
    Q1_price = df['UnitPrice'].quantile(0.25)
    Q3_price = df['UnitPrice'].quantile(0.75)
    IQR_price = Q3_price - Q1_price
    limite_inferior_price = Q1_price - 1.5 * IQR_price
    limite_superior_price = Q3_price + 1.5 * IQR_price

    # Identificar o número de outliers
    outliers_qty = df[(df['Quantity'] < limite_inferior_qty) | (df['Quantity'] > limite_superior_qty)]
    outliers_price = df[(df['UnitPrice'] < limite_inferior_price) | (df['UnitPrice'] > limite_superior_price)]

    print(f"\nNúmero de outliers em Quantity: {len(outliers_qty)}")
    print(f"Número de outliers em UnitPrice: {len(outliers_price)}")
```

A escolha de como tratar os outliers identificados depende do objetivo da análise. Não há uma única resposta correta.

| Estratégia | Descrição | Prós | Contras |
| :--- | :--- | :--- | :--- |
| **Remoção** | Excluir as linhas que contêm outliers. | Simples de implementar. Eficaz se os outliers são erros claros. | Pode levar à perda de informações valiosas. Reduz o tamanho do dataset. Pode introduzir viés se os outliers não forem aleatórios. |
| **Capping (Limitação)** | Substituir os outliers pelos valores do limite superior/inferior (ex: 99º percentil). | Mantém o tamanho do dataset. Reduz a influência de valores extremos. | É uma alteração artificial dos dados. A escolha do limite é subjetiva. |
| **Transformação** | Aplicar uma função matemática (ex: logaritmo) para reduzir a assimetria da distribuição. | Pode normalizar a distribuição dos dados, o que é um requisito para alguns modelos. | Torna a interpretação dos valores menos direta. Não funciona para valores negativos ou zero (no caso do log). |
| **Imputação** | Substituir o outlier por uma medida de tendência central, como a mediana. | Mantém a linha, perdendo apenas o valor extremo. | Pode subestimar a variabilidade real dos dados. |

Para nossa análise, vamos optar por remover os outliers para focar no comportamento "típico" dos clientes.

```python
if df is not None:
    # Remover os outliers
    df_limpo = df[(df['Quantity'] >= limite_inferior_qty) & (df['Quantity'] <= limite_superior_qty)]
    df_limpo = df_limpo[(df_limpo['UnitPrice'] >= limite_inferior_price) & (df_limpo['UnitPrice'] <= limite_superior_price)]

    print(f"\nDimensões finais do DataFrame limpo: {df_limpo.shape}")
```

## Seção 6: Primeiros Insights: Estatística Descritiva Essencial

Com os dados limpos, podemos começar a extrair os primeiros insights usando estatística descritiva.

### 6.1. Medindo o Mundo: Escalas de Medida

Em estatística, nem todos os dados são criados iguais. As **escalas de medida** definem a natureza das nossas variáveis e, crucialmente, determinam quais operações matemáticas e estatísticas são válidas para elas.

* **Nominal:** Variáveis categóricas sem uma ordem intrínseca. Ex: `Country`. Podemos contar a frequência de cada país, mas não faz sentido dizer que "França" é maior que "Alemanha".
* **Ordinal:** Variáveis categóricas com uma ordem clara, mas com intervalos não definidos. Ex: uma avaliação de produto ('ruim', 'médio', 'bom'). Sabemos que 'bom' > 'médio', mas não o quanto.
* **Intervalar:** Variáveis numéricas onde os intervalos entre os valores são iguais, mas não há um "zero absoluto". Ex: Temperatura em Celsius. A diferença entre 10°C e 20°C é a mesma que entre 20°C e 30°C, mas 20°C não é "o dobro de calor" que 10°C.
* **De Razão (Ratio):** Variáveis numéricas com intervalos iguais e um zero absoluto significativo. Ex: `Quantity` e `UnitPrice`. Um preço de R$0 significa ausência de custo, e um preço de R$20 é o dobro de R$10. Aqui, todas as operações aritméticas são permitidas.

### 6.2. Uma ou Várias Variáveis? Análise Univariada vs. Multivariada

A análise de dados pode ser dividida com base no número de variáveis que examinamos simultaneamente.

* **Análise Univariada:** Foca em analisar **uma variável por vez** para entender suas características e distribuição. Exemplos de perguntas: "Qual a média de preço dos produtos?", "Qual país tem o maior número de transações?".
* **Análise Bivariada/Multivariada:** Examina a **relação entre duas ou mais variáveis**. Exemplos de perguntas: "Existe uma correlação entre o preço do produto e a quantidade vendida?", "Clientes de qual país gastam mais, em média?".

Hoje, focaremos na análise univariada para resumir nosso dataset limpo.

### 6.3. Resumindo os Dados com Pandas (Análise Univariada na Prática)

O Pandas oferece funções diretas e poderosas para calcular as principais estatísticas descritivas.

* **`describe()`: O Resumo Completo**
    A função `describe()` é o ponto de partida para qualquer análise, fornecendo um resumo estatístico completo das colunas numéricas.

```python
if 'df_limpo' in locals():
    # Estatísticas descritivas do DataFrame limpo
    print("\nEstatísticas Descritivas do DataFrame Limpo:")
    display(df_limpo.describe())
```

A saída nos dá: `count` (contagem), `mean` (média), `std` (desvio padrão), `min` (mínimo), `25%` (primeiro quartil), `50%` (mediana), `75%` (terceiro quartil) e `max` (máximo).

* **Medidas de Tendência Central**
    Podemos calcular individualmente as medidas que indicam o "centro" de uma distribuição.

```python
if 'df_limpo' in locals():
    # Média do preço unitário
    media_preco = df_limpo['UnitPrice'].mean()
    print(f"\nMédia do Preço Unitário: ${media_preco:.2f}")

    # Mediana do preço unitário (mais robusta a outliers)
    mediana_preco = df_limpo['UnitPrice'].median()
    print(f"Mediana do Preço Unitário: ${mediana_preco:.2f}")

    # Moda do país (o país que aparece com mais frequência)
    moda_pais = df_limpo['Country'].mode()[0] # .mode() retorna uma Series, pegamos o primeiro elemento
    print(f"País com mais transações (Moda): {moda_pais}")
```

* **Análise de Frequência para Variáveis Categóricas**
    Para variáveis nominais como `Country`, a contagem de frequência é a análise mais informativa. O método `value_counts()` é perfeito para isso.

```python
if 'df_limpo' in locals():
    # Contagem de transações por país
    print("\nContagem de Transações por País:")
    display(df_limpo['Country'].value_counts())

    # Contagem de transações por país em percentual
    print("\nPercentual de Transações por País (Top 5):")
    display(df_limpo['Country'].value_counts(normalize=True).head(5))
```

O parâmetro `normalize=True` é especialmente útil para entender a proporção de cada categoria em relação ao total.
